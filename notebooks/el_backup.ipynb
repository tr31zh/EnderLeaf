```python
from pathlib import Path
import shutil

from tqdm import tqdm
from rich.pretty import pprint
from joblib import Parallel, delayed

import numpy as np
import pandas as pd
import cv2

from enderleaf.image import to_pil
```

```python
LOCAL_ROOT = Path(".").joinpath("output")
# DIST_ROOT = Path("/").joinpath("Volumes", "LaCie", "enderleaf", "output")
DIST_ROOT = Path("/").joinpath("Volumes", "Samsung_T5", "Enderleaf", "output")

[p.is_dir() for p in [LOCAL_ROOT, DIST_ROOT]]
```

```python
src_csv_files = [
    c for c in LOCAL_ROOT.rglob("*.csv") if c.name.startswith("._") is False
]
src_img_files = [
    c for c in LOCAL_ROOT.rglob("*.png") if c.name.startswith("._") is False
]

pprint(len(src_csv_files))
pprint(len(src_img_files))
```

```python
def copy_if_missing(file_path):
    dst_path = DIST_ROOT.joinpath(file_path.relative_to(LOCAL_ROOT))
    if dst_path.is_file() is False:
        shutil.copy(file_path, dst_path)
        return 1
    try:
        to_pil(cv2.imread(file_path))
    except:
        print(f"Error on {file_path}", flush=True)
        shutil.copy(file_path, dst_path)
        return 1
    return 0
```

```python
data = Parallel(n_jobs=10)(
    delayed(copy_if_missing)(file_path) for file_path in tqdm(src_csv_files)
)
np.asarray(data).sum()
```

```python
data = Parallel(n_jobs=10)(
    delayed(copy_if_missing)(file_path) for file_path in tqdm(src_img_files)
)
np.asarray(data).sum()
```
