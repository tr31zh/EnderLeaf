## Enderscopy stage control panel demo

This notebook shows how to create a stage control panel.

```python
# Needed to import from the enderscope library

import os

os.chdir("..")
```

```python
from enderscope.serial import list_ports, Stage
from enderscope.ui import Panel
```

```python
# list available serial ports
stage_port = None
light_port = None
ports = list_ports()
for port in ports:
    if "USB Serial" in port.description:
        stage_port = port
        print("* " + str(port))
    elif "UART" in port.description:
        light_port = port
        print("+ " + str(port))
    else:
        print("  " + str(port))
```

```python
# connect to enderscope and create a stage object
s = Stage(stage_port.device, 115200)
```

```python
# create a stage control panel
p = Panel(s)
```

```python
p.recorded_positions
```

```python
# iterate over manually recorded positions
for i in range(5):
    for pos in p.recorded_positions:
        s.move_position(pos)
        s.finish_moves()
        # add camera snap here
```
