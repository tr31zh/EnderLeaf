# Enderlights demo 
Direct control through Raspberry Pi 5

This notebook show how to connect and drive an enderscope illumination from Python without using an Arduino board

```python
%load_ext autoreload
%autoreload 2
```

```python
# Needed to import from the enderscope library

import os

os.chdir("..")
```

```python
from time import sleep

from enderscope.enderlights_pi import Enderlights, cycle, CardPoint
```

```python
lights = Enderlights()
```

```python
lights.shutter(True, (20, 20, 20))
```

```python
lights.fill((50, 0, 50))
```

```python
# blinking LEDs ;-)
for i in range(10):
    lights.shutter(True, (50, 0, 50))
    sleep(0.05)
    lights.shutter(False)
    sleep(0.05)
```

```python
lights.led_data
```

```python
# Cycle 1 LED
for _ in range(2):
    for i in cycle(lights=lights, step=1, size=1, value=(50, 0, 50)):
        sleep(0.5)
lights.shutter(False)
```

```python
# Cycle 3 LEDs
for _ in range(2):
    for i in cycle(lights=lights, step=4, size=4, value=(50, 0, 50)):
        sleep(0.5)
lights.shutter(False)
```

```python
# Use cardinal points to display patterns

for cps in [
    [CardPoint.NORTH],
    [CardPoint.EAST],
    [CardPoint.SOUTH],
    [CardPoint.WEST],
    [CardPoint.NORTH, CardPoint.EAST],
    [CardPoint.EAST, CardPoint.SOUTH],
    [CardPoint.SOUTH, CardPoint.WEST],
    [CardPoint.WEST, CardPoint.NORTH],
    [CardPoint.NORTH, CardPoint.SOUTH],
    [CardPoint.EAST, CardPoint.WEST],
]:
    lights.shutter(False)
    for cp in cps:
        lights.set_cardinal(cp, (50, 0, 50))
    sleep(0.5)
lights.shutter(False)
```
