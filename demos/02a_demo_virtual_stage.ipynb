## Enderscopy stage demo
### shows how to create a virtual enderscope stage

```python
# Needed to import from the enderscope library

import os

os.chdir("..")
```

```python

import time
from enderscope.serial import Stage
# connect to a virtual enderscope and create a stage object
%matplotlib widget
s = Stage('virtual', 115200)
```

```python
from enderscope.scan_patterns import snake
s.home()
origin = [25,200] # coordinates of first position
steps = [30,-25] # x and y shifts to next item
positions = snake(8,6) * steps + origin
for p in positions:
    s.move_position(p)
    
```

```python
from enderscope.ui import Panel

p = Panel(s)
```
