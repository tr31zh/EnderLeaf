```python
# Needed to import from the enderscope library

import os

os.chdir("..")
```

```python
from enderscope.scan_patterns import raster, snake, rnd_mvt, spiral, plot_path

origin = [25, 100]  # coordinates of first position
steps = [38, -25]  # x and y shifts to next item
positions = raster(4, 3) * steps + origin
plot_path(positions, field=(40, 30), title="raster scan")
```

```python
positions = snake(4, 4) * [33.33, -25] + [50, 100]
plot_path(positions, title="snake scan")
```

```python
positions = rnd_mvt(20)
plot_path(positions, title="random scan")
```

```python
positions = spiral(16) * [30, 20] + [100, 75]
plot_path(positions, field=(34, 24), title="spiral scan")
```
