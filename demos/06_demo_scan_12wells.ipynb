```python
# list available serial ports
stage_port = None
light_port = None
ports = list_ports()
for port in ports:
    if "USB Serial" in port.description:
        stage_port = port
        print("* " + str(port))
    elif "UART" in port.description:
        light_port = port
        print("+ " + str(port))
    else:
        print("  " + str(port))
```

```python
# connect to enderscope and create a stage object
s = Stage(stage_port, 115200)
l = Enderlights(light_port, 57600)
s.home()
s.move_relative(0, 0, 10)
```

```python
# create an array of positions to perform a raster scan
# import time
origin = [56, 100]  # coordinates of first position
steps = [25, -25]  # x and y offsets to next item
grid = np.array(list((x, y) for y in range(3) for x in range(4))) * steps + origin
for p in grid:
    s.move_position(p)
    s.finish_moves()
    l.shutter(True)
    time.sleep(0.5)
    l.shutter(False)

s.move_position(origin)  # move back to origin
```
